```python
import string 
import random 

is_valid=False
while is_valid==False:
    try:
        password_number=int(input("How many passwords do you want? "))
        if password_number>0:
            is_valid=True
        else:
            print("Invalid Input! Please enter a positive number. ")
    except ValueError:
        print("Invalid Input! Please enter an integer. ")

is_valid=False
while is_valid==False:
    try:
        minimum_length=int(input("How many characters, at a minimum, would you like your password to have (at least 6)? "))
        if minimum_length>=6:
            is_valid=True
        else:
            print("Invalid Input! The least password length is 6. Please try again. ")
    except ValueError:
        print("Invalid Input! Please enter an integer. ")

is_valid=False
while is_valid==False:
    try:
        maximum_length=int(input("How many characters, at a maximum, would you like your password to have? "))
        if maximum_length>=minimum_length:
            is_valid=True
        else:
            print("Invalid Input! The maximum length of password should be a number greater or egual to the minimum length. Please try again. ")
    except ValueError:
        print("Invalid Input! Please enter an integer. ")
 
is_continue=True
while is_continue==True:
    characters=[]
    lowercase=input("Do you want lowercase letters in your password? ").lower()
    while lowercase!="yes" and lowercase!="no":
        print("Invalid Input! Please enter yes or no. ")
        lowercase=input("Do you want lowercase letters in your password? ").lower()
    uppercase=input("Do you want uppercase letters in your password? ").lower()
    while uppercase!="yes" and uppercase!="no":
        print("Invalid Input! Please enter yes or no. ")
        uppercase=input("Do you want uppercase letters in your password? ").lower()
    digits=input("Do you want digits in your password? ").lower()
    while digits!="yes" and digits!="no":
        print("Invalid Input! Please enter yes or no. ")
        digits=input("Do you want digits in your password? ").lower()
    symbols=input("Do you want symbols in your password? ").lower()
    while symbols!="yes" and symbols!="no":
        print("Invalid Input! Please enter yes or no. ")
        symbols=input("Do you want symbols in your password? ").lower()
    if lowercase=="yes":
        characters.append(string.ascii_lowercase)
    if uppercase=="yes":
        characters.append(string.ascii_uppercase)
    if digits=="yes":
        characters.append(string.digits)
    if symbols=="yes":
        characters.append(string.punctuation)
    if lowercase=="no" and uppercase=="no" and digits=="no" and symbols=="no":
        print("You must choose at least one type of character. Try again please. ")
    else:
        is_continue=False 

character_pool="".join(characters)

with open("E:\Python\Password creator\Password.txt",mode="w") as file:
    for i in range(password_number):
        password_length=random.randint(minimum_length,maximum_length)
        password=""
        
        for j in range(password_length):
            password+=random.choice(character_pool)
        file.write(password+"\n")
```
